```python
#Weekly Contest 470 
#Q1. Compute Alternating Sum 
#You are given an integer array nums. The alternating sum of an array is defined as the sum of the elements at even indices minus the sum of the elements at odd indices.
#Return the alternating sum of nums.
def alternatingSum(nums):
    even_sum = sum(nums[i] for i in range(0, len(nums), 2))
    odd_sum = sum(nums[i] for i in range(1, len(nums), 2))
    return even_sum - odd_sum
#Example 1:
print(alternatingSum([4,2,5,3]))  # Output: 4
#Explanation: The alternating sum is (4 + 5) - (2 + 3) = 4.
#Example 2:
print(alternatingSum([5,6,7,8]))  # Output: -2
#Explanation: The alternating sum is (5 + 7) - (6 + 8) = -2.
```

```python
#Q2. Longest Subsequence with Non-Zero Bitwise AND
#You are given an integer array nums. 
# Return the length of the longest subsequence in nums whose bitwise XOR is non-zero. If no such subsequence exists, return 0.
def longestSubsequence(nums):
        xor_total = 0
        for num in nums:
            xor_total ^= num
        
        # If total XOR is non-zero, all elements work
        if xor_total != 0:
            return len(nums)
        
        # If all numbers are zero, no valid subsequence
        if all(num == 0 for num in nums):
            return 0
        
        # Otherwise, remove one element
        return len(nums) - 1

#Example 1:
print(longestSubsequence([3,2,1,4]))  # Output: 4
#Explanation: The whole array is a valid subsequence with a bitwise XOR of 4.
#Example 2:
print(longestSubsequence([0,0,0]))  # Output: 0
#Explanation: Every subsequence has a bitwise XOR of 0. Therefore, there is no valid subsequence.
#Example 3:
print(longestSubsequence([0,1,1,1,1]))  # Output: 4
#Explanation: The longest valid subsequence is [1,1,1,1] with a bitwise XOR of 0.

```

```python
#Q3. Remove K-Balanced Substrings 
#You are given a string s consisting of '(' and ')' only, and an integer k.
#A string is called k-balanced if:
#It is a non-empty string.
#The number of '(' characters is equal to the number of ')' characters.
#The number of '(' characters in every prefix of the string is at most k times the number of ')' characters in the same prefix.

#For example, "(()())" is 2-balanced, but "((()))" is not k-balanced for any k since the first prefix has 3 '(' characters and 0 ')' characters.
#You can remove some (possibly none) characters from s to make it k-balanced.
#Return the final string after all possible removals. 

class Solution:
    def removeSubstring(self, s: str, k: int) -> str:
        stack = []  # each item: [char, count]

        for ch in s:
            if stack and stack[-1][0] == ch:
                stack[-1][1] += 1
            else:
                stack.append([ch, 1])

            # After each push/merge, greedily remove all possible k-balanced blocks
            while len(stack) >= 2 and stack[-2][0] == '(' and stack[-1][0] == ')':
                a = stack[-2][1]  # '(' run length
                b = stack[-1][1]  # ')' run length
                t = min(a // k, b // k)  # how many k-balanced chunks we can remove
                if t == 0:
                    break

                stack[-2][1] -= t * k
                stack[-1][1] -= t * k

                # remove empty groups
                if stack[-1][1] == 0:
                    stack.pop()
                if stack and stack[-1][1] == 0:
                    stack.pop()

                # if same chars become adjacent, merge runs
                if len(stack) >= 2 and stack[-2][0] == stack[-1][0]:
                    stack[-2][1] += stack[-1][1]
                    stack.pop()

        # rebuild result
        return ''.join(ch * cnt for ch, cnt in stack)

#Example 1:
print(Solution().removeSubstring("(()())", 2))  # Output: "(()())"
#Explanation: The string is already 2-balanced, so no characters need to be removed.
#Example 2:
print(Solution().removeSubstring("((()))", 2))  # Output: "()"
#Explanation: The first prefix "(((" has 3 '(' characters and 0 ')' characters.
#We can remove 2 '(' characters to make the string 2-balanced. The resulting string
#is "()".
#Example 3:
print(Solution().removeSubstring("))())(", 1))  # Output: ""
#Explanation: We can remove all characters to make the string 1-balanced. The resulting string is "".
```

```python
#Q4. Count No-Zero Pairs That Sum to N 
#A no-zero integer is a positive integer that does not contain any 0 in its decimal representation.
#Given an integer n, count the number of pairs(a,b) where:
#1 <= a, b <= n
#a + b = n
from functools import lru_cache

class Solution:
    def countNoZeroPairs(self, n: int) -> int:
        ds = list(map(int, str(n)))[::-1]  # digits of n, least-significant first

        @lru_cache(None)
        def dp(i: int, carry: int, fa: int, fb: int) -> int:
            # i processed digits [0..i-1]; fa/fb: 1 = already finished (only zeros above), 0 = still digits to place
            if i == len(ds):
                return 1 if carry == 0 and fa == 1 and fb == 1 else 0

            need = ds[i]
            total = 0

            # choose digit x for a
            if fa:       # already finished -> only 0 allowed
                xs = [(0, 1)]  # (x, nfa=1)
            else:         # still placing digits -> 1..9, and choose whether to finish after this digit
                xs = [(x, 0) for x in range(1, 10)] + [(x, 1) for x in range(1, 10)]

            # choose digit y for b
            if fb:
                ys = [(0, 1)]
            else:
                ys = [(y, 0) for y in range(1, 10)] + [(y, 1) for y in range(1, 10)]

            for x, nfa in xs:
                for y, nfb in ys:
                    s = x + y + carry
                    if s % 10 == need:
                        total += dp(i + 1, s // 10, nfa, nfb)

            return total

        return dp(0, 0, 0, 0)
#Example 1:
print(Solution().countNoZeroPairs(20))  # Output: 19
#Explanation: There are 19 valid pairs (a, b) where a + b = 20:
#(1, 19), (2, 18), (3, 17), (4, 16), (5, 15), (6, 14), (7, 13), (8, 12), (9, 11),
#(11, 9), (12, 8), (13, 7), (14, 6), (15, 5), (16, 4), (17, 3), (18, 2), (19, 1).
#Example 2:
print(Solution().countNoZeroPairs(11))  # Output: 10
#Explanation: There are 10 valid pairs (a, b) where a + b = 11:
#(1, 10), (2, 9), (3, 8), (4, 7), (5, 6), (6, 5), (7, 4), (8, 3), (9, 2), (10, 1).


```
