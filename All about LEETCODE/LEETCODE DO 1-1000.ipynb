```python
#1. Two Sum 
#Given an array of integers nums and an integer target, return indices of the two numbers such that they add to target. 
#You may assume that each input would have exactly one solution, and you may not use the the same element twice. 

def twoSum(nums, target): 
    num_to_index = {}
    for i, num in enumerate(nums): 
        complement = target - num # Find the complement that adds up to the target
        if complement in num_to_index: 
            return [num_to_index[complement], i] # Return the indices of the two numbers
        num_to_index[num] = i #Store the index of the current number 
    return [] # Return an empty list if no solution is found
    
# Example usage
nums = [2, 7, 11, 15]
target = 9
result = twoSum(nums, target)
print(result)  # Output: [0, 1]

```

```python
#3. Longest Substring Without Repeating Characters
# Given a string s, find the length of the longest substring without repeating characters.
def lengthOfLongestSubstring(s): 
    char_index_map = {} # Dictionary to store the last index of each character
    left = 0 # Left pointer of the sliding window
    max_length = 0 # Variable to store the maximum length found

    for right, char in enumerate(s): # Iterate over the string with the right pointer
        if char in char_index_map and char_index_map[char] >= left: 
            left = char_index_map[char] + 1 # Move the left pointer to the right of the last occurrence
        char_index_map[char] = right # Update the last index of the character
        max_length = max(max_length, right - left + 1) # Update the maximum length found
    return max_length # Return the maximum length of substring found
# Example usage
s = "abcabcbb"
result = lengthOfLongestSubstring(s)
print(result)  # Output: 3 (The answer is "abc", with the length of 3)

```

```python
#4. Median of Two Sorted Arrays
# Given two sorted arrays nums1 and nums2 of size m and n respectively, return the median of the two sorted arrays.
# The overall run time complexity should be O(log (m+n)).
def findMedianSortedArrays(nums1, nums2):
    # Ensure nums1 is the smaller array 
    if len(nums1) > len(nums2):
        nums1, nums2 = nums2, nums1
    x, y = len(nums1), len(nums2)
    low, high = 0, x

    while low <= high:
        partitionX = (low + high) // 2
        partitionY = (x+y+1) // 2 - partitionX 
        maxX = float('-inf') if partitionX == 0 else nums1[partitionX - 1] # Handle edge cases
        maxY = float('-inf') if partitionY == 0 else nums2[partitionY - 1] # Handle edge cases
        minX = float('inf') if partitionX == x else nums1[partitionX] # Handle edge cases
        minY = float('inf') if partitionY == y else nums2[partitionY] # Handle edge cases
        if maxX <= minY and maxY <= minX: 
            if (x + y) % 2 == 0: 
                return (max(maxX, maxY) + min(minX, minY)) / 2 # If even, return the average of the two middle values
            else:
                return max(maxX, maxY) # If odd, return the middle value
        elif maxX > minY:
            high = partitionX - 1 # Move towards the left in nums1
        else:
            low = partitionX + 1 # Move towards the right in nums1
# Example usage
nums1 = [1, 3]
nums2 = [2]
result = findMedianSortedArrays(nums1, nums2)
print(result)  # Output: 2.0


    
```

```python
#5. Longest Palindromic Substring
# Given a string s, return the longest palindromic substring in s.
def longestPalindrome(s): 
        if len(s) < 2:
            return s

        start, max_length = 0, 1

        for i in range(len(s)): # Iterate through each character in the string
            left, right = i - 1, i + 1
            while right < len(s) and s[right] == s[i]: # Handle even-length palindromes
                right += 1
            while left >= 0 and s[left] == s[i]: # Handle odd-length palindromes
                left -= 1
            while left >= 0 and right < len(s) and s[left] == s[right]: # Expand around the center
                left -= 1
                right += 1
            length = right - left - 1
            if length > max_length:
                max_length = length
                start = left + 1

        return s[start:start + max_length]
# Example usage
s = "babad"
result = longestPalindrome(s)
print(result)  # Output: "bab" or "aba"
            
```

```python
#6. ZigZag Conversion
# The string "PAYPALISHIRING" is written in a zigzag pattern on a given number of rows like this: (you may want to display this pattern in a fixed font for better legibility)
# P   A   H   N
# A P L S I I G
# Y   I   R
# And then read line by line: "PAHNAPLSIIGYIR"
def convert(s, numRows): 
        if numRows == 1 or numRows >= len(s):
            return s
        
        rows = [''] * numRows
        current_row = 0
        going_down = False

        for char in s:
            rows[current_row] += char
            if current_row == 0 or current_row == numRows - 1:
                going_down = not going_down
            current_row += 1 if going_down else -1
        
        return ''.join(rows)
            
# Example usage
s = "PAYPALISHIRING"
numRows = 3
result = convert(s, numRows)
print(result)  # Output: "PAHNAPLSIIGYIR"
# Example usage
s = "PAYPALISHIRING"
numRows = 4
result = convert(s, numRows)
print(result)  # Output: "PINALSIGYA
```

```python
#7. Reverse Integer
# Given a signed 32-bit integer x, return x with its digits reversed. If reversing x causes the value to go outside the signed 32-bit integer range [-231, 231 - 1], then return 0.
def reverse(x): 
    INT_MIN, INT_MAX = -2**31, 2**31 - 1 #range: [-2,147,483,648 to 2,147,483,647]
    sign = -1 if x < 0 else 1 # Determine the sign of the integer
    x *= sign # Make x positive for easier reversal
    reversed_x = 0
    while x: 
        digit = x % 10 # Get the last digit
        x //= 10 # Remove the last digit from x
        reversed_x = reversed_x * 10 + digit # Append the digit to the reversed number
        if reversed_x > INT_MAX: 
            return 0 # Return 0 if it goes out of bounds
    return sign * reversed_x # Restore the original sign
# Example usage
x = 123
result = reverse(x)
print(result)  # Output: 321
# Example usage
x = -123
result = reverse(x)
print(result)  # Output: -321


    
```

```python
#8. String to Integer (atoi)
# Implement the myAtoi(string s) function, which converts a string to a 32-bit signed integer (similar to C/C++'s atoi function).
# The algorithm for myAtoi(string s) is as follows:
# 1. whitespace: ignore leading whitespace (" ")
# 2. signedness: check if the next character (if not already at the end of the string) is '-' or '+'. Read this character in if it is either. This determines if the final result is negative or positive respectively. Assume the result is positive if neither is present.
# 3. conversion: read the integer by skipping leading zeros until a non-digit character or the end of the input is reached. The rest of the string is ignored.
# 4. rounding: if the integer is out of the 32-bit signed integer range [-2^31, 2^31 - 1], clamp the integer so that it remains in the range. Specifically, integers less than -2^31 should be clamped to -2^31, and integers greater than 2^31 - 1 should be clamped to 2^31 - 1.
def myAtoi(s):
    INT_MIN, INT_MAX = -2**31, 2**31 - 1 #case edge 
    i, n, sign = 0, len(s), 1
    result = 0 

    #1. skip leading whitespaces 
    while i < n and s[i] == " ":
        i += 1
    #2. check for sign
    if i < n and (s[i] == "-" or s[i] == "+"):
        sign = -1 if s[i] == "-" else 1
        i += 1

    #3. convert digits to integer
    while i < n and s[i] .isdigit():
        digit = int(s[i])
        if result > (INT_MAX- digit) // 10: # Check for overflow
            return INT_MAX if sign == 1 else INT_MIN
        result = result * 10 + digit
        i += 1

    return sign * result # Apply the sign to the result
# Example usage
s = "   -42"
print(myAtoi(s))  # Output: -42
```

```python
#9. Palindrome Number
# Given an integer x, return true if x is a palindrome, and false otherwise.

def isPalindrome(x): 
    if x < 0: 
        return False # Negative numbers are not palindromes
    original, reversed_num = x, 0 
    while x > 0: 
        digit = x % 10 # Get the last digit
        reversed_num = reversed_num * 10 + digit 
        x //= 10 # Remove the last digit from x
    return original == reversed_num # Check if the original number is equal to the reversed number
# Example usage
x = 121
result = isPalindrome(x)
print(result)  # Output: True
# Example usage
x = -121
result = isPalindrome(x)
print(result)  # Output: False
```

```python
#10. Regular Expression Matching 
# Given an input string s and a pattern p, implement regular expression matching with support for '.' and '*' where:
# '.' Matches any single character.​​​​
# '*' Matches zero or more of the preceding element.
# The matching should cover the entire input string (not partial).

def isMatch(s, p): 
    m, n = len(s), len(p)
    dp = [[False] * (n + 1) for _ in range(m + 1)] # Create a 2D DP table
    dp[0][0] = True # Empty string matches empty pattern
    for j in range(1, n + 1): 
        if p[j - 1] == '*': 
            dp[0][j] = dp[0][j - 2] # '*' can match zero occurrences of the preceding element
    for i in range(1, m + 1): 
        for j in range(1, n + 1): 
            if p[j - 1] == s[i - 1] or p[j - 1] == '.': 
                dp[i][j] = dp[i - 1][j - 1] # Characters match
            elif p[j - 1] == '*': 
                dp[i][j] = dp[i][j - 2] # '*' matches zero occurrences
                if p[j - 2] == s[i - 1] or p[j - 2] == '.': 
                    dp[i][j] = dp[i][j] or dp[i - 1][j] # '*' matches one or more occurrences
    return dp[m][n] # Return whether the entire string matches the entire pattern
# Example usage
s = "aab"
p = "c*a*b"
result = isMatch(s, p)
print(result)  # Output: True
# Example usage
s = "mississippi"
p = "mis*is*p*."
result = isMatch(s, p)
print(result)  # Output: False
```

```python
#11. Container With Most Water
# You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).
# Find two lines that together with the x-axis form a container, such that the container contains the most water.
def maxArea(height): 
    left, right = 0, len(height) - 1 # Initialize two pointers
    max_area = 0 # Variable to store the maximum area found
    while left < right:
        width = right - left # Calculate the width between the two pointers
        current_height = min(height[left], height[right]) # Height is determined by the shorter line
        current_area = width * current_height # Calculate the area
        max_area = max(max_area, current_area) # Update the maximum area found
        if height[left] < height[right]: 
            left += 1 # Move the left pointer to the right
        else:
            right -= 1 # Move the right pointer to the left
    return max_area # Return the maximum area found
# Example usage
height = [1,8,6,2,5,4,8,3,7]
result = maxArea(height)
print(result)  # Output: 49
# Example usage
height = [1,1]
result = maxArea(height)
print(result)  # Output: 1
```

```python
#12. Integer to Roman
# Roman numerals are represented by seven different symbols: I, V, X, L, C, D and M.
# Symbol       Value
# I             1
# V             5
# X             10
# L             50
# C             100
# D             500
# M             1000
def intToRoman(num): 
    val = [
        1000, 900, 500, 400,
        100, 90, 50, 40,
        10, 9, 5, 4,
        1 # Values of Roman numeral symbols, including subtractive combinations
    ]
    syms = [
        "M", "CM", "D", "CD",
        "C", "XC", "L", "XL",
        "X", "IX", "V", "IV",
        "I" # Corresponding Roman numeral symbols
    ]
    roman_numeral = ""
    for i in range(len(val)): 
        while num >= val[i]: 
            num -= val[i] # Subtract the value from num
            roman_numeral += syms[i] # Append the corresponding symbol to the result
    return roman_numeral # Return the final Roman numeral string
# Example usage
num = 1994
result = intToRoman(num)
print(result)  # Output: "MCMXCIV"
# Example usage
num = 58
result = intToRoman(num)
print(result)  # Output: "LVIII"
```

```python
#14. Longest Common Prefix
# Write a function to find the longest common prefix string amongst an array of strings.
def longestCommonPrefix(strs): 
    if not strs: 
        return "" # Return empty string if the input list is empty
    prefix = strs[0] # Start with the first string as the initial prefix
    for s in strs[1:]: 
        while not s.startswith(prefix): 
            prefix = prefix[:-1] # Shorten the prefix until it matches the start of the current string
            if not prefix: 
                return "" # Return empty string if no common prefix is found
    return prefix # Return the longest common prefix found

# Example usage
strs = ["flower", "flow", "flight"]
result = longestCommonPrefix(strs)
print(result)  # Output: "fl"
# Example usage
strs = ["dog", "racecar", "car"]
result = longestCommonPrefix(strs)
print(result)  # Output: ""
```

```python
#Daily Challenge - 9 October 2025
#Find the Minimum Amount of Time to Brew Potions
# You are given two integer arrays, skill and mana, of length n and m, respectively.

# In a laboratory, n wizards must brew m potions in order. Each potion has a mana capacity mana[j] and must pass through all the wizards sequentially to be brewed properly. The time taken by the ith wizard on the jth potion is timeij = skill[i] * mana[j].

# Since the brewing process is delicate, a potion must be passed to the next wizard immediately after the current wizard completes their work. This means the timing must be synchronized so that each wizard begins working on a potion exactly when it arrives. ​

# Return the minimum amount of time required for the potions to be brewed properly.

from typing import List

class Solution:
    def minTime(self, skill: List[int], mana: List[int]) -> int:
        n = len(skill)
        m = len(mana)

        # Prefix sums of skills: S[0]=0, S[r]=sum_{i<=r} skill[i-1]
        S = [0] * (n + 1)
        for r in range(1, n + 1):
            S[r] = S[r - 1] + skill[r - 1]

        # Build points (x, y) = (S_r, a_r) with x increasing
        pts = [(S[r], skill[r - 1]) for r in range(1, n + 1)]

        # Upper convex hull (monotone chain) for max dot with (vx, vy) where vy >= 0
        def cross(o, a, b):
            return (a[0] - o[0]) * (b[1] - o[1]) - (a[1] - o[1]) * (b[0] - o[0])

        hull = []
        for p in pts:
            while len(hull) >= 2 and cross(hull[-2], hull[-1], p) >= 0:
                hull.pop()
            hull.append(p)

        # Precompute edge slopes (strictly decreasing on the upper hull)
        slopes = []
        for i in range(len(hull) - 1):
            dx = hull[i + 1][0] - hull[i][0]
            dy = hull[i + 1][1] - hull[i][1]
            slopes.append(dy / dx)

        # Query max dot (vx, vy) · (x, y) on the hull; here vy = next_mana >= 0
        def max_dot(vx: int, vy: int) -> int:
            if len(hull) == 1:
                x, y = hull[0]
                return vx * x + vy * y
            if vy == 0:
                # maximize vx * x: pick extreme x depending on sign
                x, y = (hull[-1] if vx >= 0 else hull[0])
                return vx * x + vy * y
            k = -vx / vy  # threshold slope

            # slopes is decreasing
            if k >= slopes[0]:
                x, y = hull[0]
                return vx * x + vy * y
            if k <= slopes[-1]:
                x, y = hull[-1]
                return vx * x + vy * y

            lo, hi = 0, len(slopes) - 1
            # Find edge index where slopes[lo-1] >= k > slopes[lo]
            while lo <= hi:
                mid = (lo + hi) // 2
                if slopes[mid] >= k:
                    lo = mid + 1
                else:
                    hi = mid - 1
            x, y = hull[lo]  # vertex after that edge
            return vx * x + vy * y

        # Total time = processing time of the last potion + sum of gaps
        total = S[n] * mana[-1]
        for j in range(m - 1):
            x, y = mana[j], mana[j + 1]
            total += max_dot(x - y, y)

        return total
# Example usage
skill = [3,1,4]
mana = [1,2,3]
solution = Solution()
print(solution.minTime(skill, mana))  # Output the minimum amount of time to brew potions
```

```python
#Daily Challenge - 10 October 2025 
#Taking Maximum Energy From the Mystic Dungeon 
# In a mystic dungeon, n magicians are standing in a line. Each magician has an attribute that gives you energy. Some magicians can give you negative energy, which means taking energy from you.

# You have been cursed in such a way that after absorbing energy from magician i, you will be instantly transported to magician (i + k). This process will be repeated until you reach the magician where (i + k) does not exist.

# In other words, you will choose a starting point and then teleport with k jumps until you reach the end of the magicians' sequence, absorbing all the energy during the journey.

# You are given an array energy and an integer k. Return the maximum possible energy you can gain.

# Note that when you are reach a magician, you must take energy from them, whether it is negative or positive energy.
from typing import List

class Solution:
    def maximumEnergy(self, energy: List[int], k: int) -> int:
        n = len(energy)
        sums = [0] * k          # suffix sums per residue class
        ans = -10**18           # or float('-inf')
        for i in range(n - 1, -1, -1):
            r = i % k
            sums[r] += energy[i]
            if sums[r] > ans:
                ans = sums[r]
        return ans

# Example usage
energy = [2, -1, 3, -4, 5]
k = 2
solution = Solution()
print(solution.maximumEnergy(energy, k))  # Output the maximum possible energy


```

```python
#Daily Challenge - 11 October 2025
from typing import List
from collections import Counter
import bisect

class Solution:
    def maximumTotalDamage(self, power: List[int]) -> int:
        freq = Counter(power)
        vals = sorted(freq.keys())                 # distinct damages
        W = [v * freq[v] for v in vals]            # weights per value

        # For each i, find last index j with vals[j] <= vals[i] - 3
        prev = []
        for v in vals:
            j = bisect.bisect_right(vals, v - 3) - 1
            prev.append(j)

        n = len(vals)
        dp = [0] * (n + 1)  # dp[i]: best using first i values (0..i-1)
        for i in range(1, n + 1):
            take = W[i-1] + (dp[prev[i-1] + 1] if prev[i-1] >= 0 else 0)
            skip = dp[i-1]
            dp[i] = max(skip, take)
        return dp[n]
# Example usage
power = [1, 4, 3, 2, 5]
solution = Solution()
print(solution.maximumTotalDamage(power))  # Output the maximum total damage
```

```python
#Daily Challenge - 12 October 2025
from typing import List

MOD = 10**9 + 7

class Solution:
    def magicalSum(self, m: int, k: int, nums: List[int]) -> int:
        # factorials for EGFs
        fact = [1] * (m + 1)
        for i in range(1, m + 1):
            fact[i] = fact[i - 1] * i % MOD
        inv_fact = [1] * (m + 1)
        inv_fact[m] = pow(fact[m], MOD - 2, MOD)
        for i in range(m, 0, -1):
            inv_fact[i - 1] = inv_fact[i] * i % MOD

        # dp[t][b][r] = coefficient after some prefix:
        # t = total picks used, b = bits so far, r = carry to next bit position
        dp = [[[0] * (m + 1) for _ in range(k + 1)] for __ in range(m + 1)]
        dp[0][0][0] = 1

        for a in nums:
            contrib = [(pow(a, c, MOD) * inv_fact[c]) % MOD for c in range(m + 1)]
            nxt = [[[0] * (m + 1) for _ in range(k + 1)] for __ in range(m + 1)]

            for t in range(m + 1):
                for b in range(k + 1):
                    for r in range(m + 1):
                        cur = dp[t][b][r]
                        if not cur:
                            continue
                        rem = m - t
                        # choose this index c times
                        for c in range(rem + 1):
                            s = r + c
                            bit = s & 1
                            nb = b + bit
                            if nb > k:
                                continue
                            nt = t + c
                            nr = s >> 1
                            if nr <= m:
                                nxt[nt][nb][nr] = (nxt[nt][nb][nr] + cur * contrib[c]) % MOD
            dp = nxt

        # finalize: remaining carry contributes popcount(carry) more ones
        ans = 0
        for b in range(k + 1):
            for r in range(m + 1):
                val = dp[m][b][r]
                if val and b + r.bit_count() == k:
                    ans = (ans + val) % MOD

        # undo the EGF scaling
        return ans * fact[m] % MOD


```

```python
print("Hello, World!")
#Daily Challenge - 13 October 2025
#Find Sum of Array Product of Magical Sequences
# You are given two integers, m and k, and an integer array nums.

# A sequence of integers seq is called magical if:
# seq has a size of m.
# 0 <= seq[i] < nums.length
# The binary representation of 2seq[0] + 2seq[1] + ... + 2seq[m - 1] has k set bits.
# The array product of this sequence is defined as prod(seq) = (nums[seq[0]] * nums[seq[1]] * ... * nums[seq[m - 1]]).

# Return the sum of the array products for all valid magical sequences.

# Since the answer may be large, return it modulo 109 + 7.

# A set bit refers to a bit in the binary representation of a number that has a value of 1.

from typing import List

MOD = 10**9 + 7

class Solution:
    def magicalSum(self, m: int, k: int, nums: List[int]) -> int:
        # factorials for EGFs
        fact = [1] * (m + 1)
        for i in range(1, m + 1):
            fact[i] = fact[i - 1] * i % MOD
        inv_fact = [1] * (m + 1)
        inv_fact[m] = pow(fact[m], MOD - 2, MOD)
        for i in range(m, 0, -1):
            inv_fact[i - 1] = inv_fact[i] * i % MOD

        # dp[t][b][r] = coefficient after some prefix:
        # t = total picks used, b = bits so far, r = carry to next bit position
        dp = [[[0] * (m + 1) for _ in range(k + 1)] for __ in range(m + 1)]
        dp[0][0][0] = 1

        for a in nums:
            contrib = [(pow(a, c, MOD) * inv_fact[c]) % MOD for c in range(m + 1)]
            nxt = [[[0] * (m + 1) for _ in range(k + 1)] for __ in range(m + 1)]

            for t in range(m + 1):
                for b in range(k + 1):
                    for r in range(m + 1):
                        cur = dp[t][b][r]
                        if not cur:
                            continue
                        rem = m - t
                        # choose this index c times
                        for c in range(rem + 1):
                            s = r + c
                            bit = s & 1
                            nb = b + bit
                            if nb > k:
                                continue
                            nt = t + c
                            nr = s >> 1
                            if nr <= m:
                                nxt[nt][nb][nr] = (nxt[nt][nb][nr] + cur * contrib[c]) % MOD
            dp = nxt

        # finalize: remaining carry contributes popcount(carry) more ones
        ans = 0
        for b in range(k + 1):
            for r in range(m + 1):
                val = dp[m][b][r]
                if val and b + r.bit_count() == k:
                    ans = (ans + val) % MOD

        # undo the EGF scaling
        return ans * fact[m] % MOD

# Example usage
m = 3
k = 2
nums = [1, 2, 3]
solution = Solution()
print(solution.magicalSum(m, k, nums))  # Output the sum of array products of magical sequences
```

```python
#Daily Challenge - 14 October 2025
#Find Resultant Array After Removing Anagrams
# You are given a 0-indexed string array words, where words[i] consists of lowercase English letters.

# In one operation, select any index i such that 0 < i < words.length and words[i - 1] and words[i] are anagrams, and delete words[i] from words. Keep performing this operation as long as you can select an index that satisfies the conditions.

# Return words after performing all operations. It can be shown that selecting the indices for each operation in any arbitrary order will lead to the same result.

# An Anagram is a word or phrase formed by rearranging the letters of a different word or phrase using all the original letters exactly once. For example, "dacb" is an anagram of "abdc".

from typing import List

class Solution:
    def removeAnagrams(self, words: List[str]) -> List[str]:
        res = []
        prev_sig = None  # signature of last kept word
        for w in words:
            sig = ''.join(sorted(w))
            if sig != prev_sig:
                res.append(w)
                prev_sig = sig
        return res
# Example usage
words = ["abba","baba","bbaa","cd","cd"]
solution = Solution()
print(solution.removeAnagrams(words))  # Output the resultant array after removing anagrams

#Adjacent Increasing Subarrays Detection 1
#  Given an array nums of n integers and an integer k, determine whether there exist two adjacent subarrays of length k such that both subarrays are strictly increasing. Specifically, check if there are two subarrays starting at indices a and b (a < b), where:
# Both subarrays nums[a..a + k - 1] and nums[b..b + k - 1] are strictly increasing.
# The subarrays must be adjacent, meaning b = a + k.
# Return true if it is possible to find two such subarrays, and false otherwise.

def hasIncreasingSubarrays(nums, k): 
    n = len(nums)
    if n < 2 * k: 
        return False # Not enough elements for two subarrays of length k

    increasing = [0] * n # Array to track lengths of increasing sequences
    for i in range(1, n): 
        if nums[i] > nums[i - 1]: 
            increasing[i] = increasing[i - 1] + 1 # Increment the length of the increasing sequence
        else:
            increasing[i] = 0 # Reset if the sequence is not increasing

    for i in range(k, n - k + 1): 
        if increasing[i - 1] >= k - 1 and increasing[i + k - 1] >= k - 1: 
            return True # Found two adjacent increasing subarrays of length k
    return False # No such subarrays found
# Example usage
nums = [1, 2, 3, 4, 5, 6]
k = 3
result = hasIncreasingSubarrays(nums, k)
print(result)  # Output: True
```

```python
#Daily Challenge - 15 October 2025
#Adjacent Increasing Subarrays Detection II

#Given an array nums of n integers, your task is to find the maximum value of k for which there exist two adjacent subarrays of length k each, such that both subarrays are strictly increasing. Specifically, check if there are two subarrays of length k starting at indices a and b (a < b), where:

    # Both subarrays nums[a..a + k - 1] and nums[b..b + k - 1] are strictly increasing.
    # The subarrays must be adjacent, meaning b = a + k.
    # Return the maximum possible value of k.

    # A subarray is a contiguous non-empty sequence of elements within an array.


from typing import List

class Solution:
    def maxIncreasingSubarrays(self, nums: List[int]) -> int:
        n = len(nums)
        if n < 2:
            return 0

        # L[i]: length of strictly increasing run ending at i
        L = [1] * n
        for i in range(1, n):
            if nums[i-1] < nums[i]:
                L[i] = L[i-1] + 1

        # R[i]: length of strictly increasing run starting at i
        R = [1] * n
        for i in range(n-2, -1, -1):
            if nums[i] < nums[i+1]:
                R[i] = R[i+1] + 1

        # Try every cut between j and j+1
        ans = 0
        for j in range(n-2, -1, -1):
            ans = max(ans, min(L[j], R[j+1]))
        return ans

# Example usage
nums = [1, 2, 3, 4, 5, 6]
solution = Solution()
print(solution.maxIncreasingSubarrays(nums))  # Output the maximum value of k for adjacent increasing subarrays

```
