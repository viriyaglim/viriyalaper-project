```python
#1. Two Sum 
#Given an array of integers nums and an integer target, return indices of the two numbers such that they add to target. 
#You may assume that each input would have exactly one solution, and you may not use the the same element twice. 

def twoSum(nums, target): 
    num_to_index = {}
    for i, num in enumerate(nums): 
        complement = target - num # Find the complement that adds up to the target
        if complement in num_to_index: 
            return [num_to_index[complement], i] # Return the indices of the two numbers
        num_to_index[num] = i #Store the index of the current number 
    return [] # Return an empty list if no solution is found
    
# Example usage
nums = [2, 7, 11, 15]
target = 9
result = twoSum(nums, target)
print(result)  # Output: [0, 1]

```

```python
#2. Add Two Numbers
# You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list. You may assume the two numbers do not contain any leading zero, except the number 0 itself.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
def addTwoNumbers(l1, l2): 
    dummy_head = ListNode(0) # Dummy head to simplify the code
    current = dummy_head # Pointer to the current node in the result list
    carry = 0 # Initialize carry to 0

    while l1 or l2 or carry: 
        val1 = l1.val if l1 else 0 #Get the value from l1 or 0 if l1 is None
        val2 = l2.val if l2 else 0 #Get the value from l2 or 0 if l2 is None
        total = val1 + val2 + carry # Calculate the sum of the two digits and the carry
        carry = total // 10 # Update carry for the next iteration
        current.next = ListNode (total % 10) # Create a new node with the digit value
        current = current.next # Move to the next node
        if l1: l1 = l1.next # Move to the next node in l1
        if l2: l2 = l2.next # Move to the next node in l2
        return dummy_head.next # Return the next node of the dummy head as the result
# Example usage
l1 = ListNode(2, ListNode(4, ListNode(3))) #
l2 = ListNode(5, ListNode(6, ListNode(4))) #
result = addTwoNumbers(l1, l2)

# Print the result
while result: 
    print(result.val, end=" -> " if result.next else "\n") 
    result = result.next


```

```python
#3. Longest Substring Without Repeating Characters
# Given a string s, find the length of the longest substring without repeating characters.
def lengthOfLongestSubstring(s): 
    char_index_map = {} # Dictionary to store the last index of each character
    left = 0 # Left pointer of the sliding window
    max_length = 0 # Variable to store the maximum length found

    for right, char in enumerate(s): # Iterate over the string with the right pointer
        if char in char_index_map and char_index_map[char] >= left: 
            left = char_index_map[char] + 1 # Move the left pointer to the right of the last occurrence
        char_index_map[char] = right # Update the last index of the character
        max_length = max(max_length, right - left + 1) # Update the maximum length found
    return max_length # Return the maximum length of substring found
# Example usage
s = "abcabcbb"
result = lengthOfLongestSubstring(s)
print(result)  # Output: 3 (The answer is "abc", with the length of 3)

```

```python
#4. Median of Two Sorted Arrays
# Given two sorted arrays nums1 and nums2 of size m and n respectively, return the median of the two sorted arrays.
# The overall run time complexity should be O(log (m+n)).
def findMedianSortedArrays(nums1, nums2):
    # Ensure nums1 is the smaller array 
    if len(nums1) > len(nums2):
        nums1, nums2 = nums2, nums1
    x, y = len(nums1), len(nums2)
    low, high = 0, x

    while low <= high:
        partitionX = (low + high) // 2
        partitionY = (x+y+1) // 2 - partitionX 
        maxX = float('-inf') if partitionX == 0 else nums1[partitionX - 1] # Handle edge cases
        maxY = float('-inf') if partitionY == 0 else nums2[partitionY - 1] # Handle edge cases
        minX = float('inf') if partitionX == x else nums1[partitionX] # Handle edge cases
        minY = float('inf') if partitionY == y else nums2[partitionY] # Handle edge cases
        if maxX <= minY and maxY <= minX: 
            if (x + y) % 2 == 0: 
                return (max(maxX, maxY) + min(minX, minY)) / 2 # If even, return the average of the two middle values
            else:
                return max(maxX, maxY) # If odd, return the middle value
        elif maxX > minY:
            high = partitionX - 1 # Move towards the left in nums1
        else:
            low = partitionX + 1 # Move towards the right in nums1
# Example usage
nums1 = [1, 3]
nums2 = [2]
result = findMedianSortedArrays(nums1, nums2)
print(result)  # Output: 2.0


    
```

```python
#5. Longest Palindromic Substring
# Given a string s, return the longest palindromic substring in s.
def longestPalindrome(s): 
        if len(s) < 2:
            return s

        start, max_length = 0, 1

        for i in range(len(s)): # Iterate through each character in the string
            left, right = i - 1, i + 1
            while right < len(s) and s[right] == s[i]: # Handle even-length palindromes
                right += 1
            while left >= 0 and s[left] == s[i]: # Handle odd-length palindromes
                left -= 1
            while left >= 0 and right < len(s) and s[left] == s[right]: # Expand around the center
                left -= 1
                right += 1
            length = right - left - 1
            if length > max_length:
                max_length = length
                start = left + 1

        return s[start:start + max_length]
# Example usage
s = "babad"
result = longestPalindrome(s)
print(result)  # Output: "bab" or "aba"
            
```

```python
#6. ZigZag Conversion
# The string "PAYPALISHIRING" is written in a zigzag pattern on a given number of rows like this: (you may want to display this pattern in a fixed font for better legibility)
# P   A   H   N
# A P L S I I G
# Y   I   R
# And then read line by line: "PAHNAPLSIIGYIR"
def convert(s, numRows): 
        if numRows == 1 or numRows >= len(s):
            return s
        
        rows = [''] * numRows
        current_row = 0
        going_down = False

        for char in s:
            rows[current_row] += char
            if current_row == 0 or current_row == numRows - 1:
                going_down = not going_down
            current_row += 1 if going_down else -1
        
        return ''.join(rows)
            
# Example usage
s = "PAYPALISHIRING"
numRows = 3
result = convert(s, numRows)
print(result)  # Output: "PAHNAPLSIIGYIR"
# Example usage
s = "PAYPALISHIRING"
numRows = 4
result = convert(s, numRows)
print(result)  # Output: "PINALSIGYA
```

```python
#7. Reverse Integer
# Given a signed 32-bit integer x, return x with its digits reversed. If reversing x causes the value to go outside the signed 32-bit integer range [-231, 231 - 1], then return 0.
def reverse(x): 
    INT_MIN, INT_MAX = -2**31, 2**31 - 1 #range: [-2,147,483,648 to 2,147,483,647]
    sign = -1 if x < 0 else 1 # Determine the sign of the integer
    x *= sign # Make x positive for easier reversal
    reversed_x = 0
    while x: 
        digit = x % 10 # Get the last digit
        x //= 10 # Remove the last digit from x
        reversed_x = reversed_x * 10 + digit # Append the digit to the reversed number
        if reversed_x > INT_MAX: 
            return 0 # Return 0 if it goes out of bounds
    return sign * reversed_x # Restore the original sign
# Example usage
x = 123
result = reverse(x)
print(result)  # Output: 321
# Example usage
x = -123
result = reverse(x)
print(result)  # Output: -321


    
```

```python
#8. String to Integer (atoi)
# Implement the myAtoi(string s) function, which converts a string to a 32-bit signed integer (similar to C/C++'s atoi function).
# The algorithm for myAtoi(string s) is as follows:
# 1. whitespace: ignore leading whitespace (" ")
# 2. signedness: check if the next character (if not already at the end of the string) is '-' or '+'. Read this character in if it is either. This determines if the final result is negative or positive respectively. Assume the result is positive if neither is present.
# 3. conversion: read the integer by skipping leading zeros until a non-digit character or the end of the input is reached. The rest of the string is ignored.
# 4. rounding: if the integer is out of the 32-bit signed integer range [-2^31, 2^31 - 1], clamp the integer so that it remains in the range. Specifically, integers less than -2^31 should be clamped to -2^31, and integers greater than 2^31 - 1 should be clamped to 2^31 - 1.
def myAtoi(s):
    INT_MIN, INT_MAX = -2**31, 2**31 - 1 #case edge 
    i, n, sign = 0, len(s), 1
    result = 0 

    #1. skip leading whitespaces 
    while i < n and s[i] == " ":
        i += 1
    #2. check for sign
    if i < n and (s[i] == "-" or s[i] == "+"):
        sign = -1 if s[i] == "-" else 1
        i += 1

    #3. convert digits to integer
    while i < n and s[i] .isdigit():
        digit = int(s[i])
        if result > (INT_MAX- digit) // 10: # Check for overflow
            return INT_MAX if sign == 1 else INT_MIN
        result = result * 10 + digit
        i += 1

    return sign * result # Apply the sign to the result
# Example usage
s = "   -42"
print(myAtoi(s))  # Output: -42
```

```python
#9. Palindrome Number
# Given an integer x, return true if x is a palindrome, and false otherwise.

def isPalindrome(x): 
    if x < 0: 
        return False # Negative numbers are not palindromes
    original, reversed_num = x, 0 
    while x > 0: 
        digit = x % 10 # Get the last digit
        reversed_num = reversed_num * 10 + digit 
        x //= 10 # Remove the last digit from x
    return original == reversed_num # Check if the original number is equal to the reversed number
# Example usage
x = 121
result = isPalindrome(x)
print(result)  # Output: True
# Example usage
x = -121
result = isPalindrome(x)
print(result)  # Output: False
```

```python
#10. Regular Expression Matching 
# Given an input string s and a pattern p, implement regular expression matching with support for '.' and '*' where:
# '.' Matches any single character.​​​​
# '*' Matches zero or more of the preceding element.
# The matching should cover the entire input string (not partial).

def isMatch(s, p): 
    m, n = len(s), len(p)
    dp = [[False] * (n + 1) for _ in range(m + 1)] # Create a 2D DP table
    dp[0][0] = True # Empty string matches empty pattern
    for j in range(1, n + 1): 
        if p[j - 1] == '*': 
            dp[0][j] = dp[0][j - 2] # '*' can match zero occurrences of the preceding element
    for i in range(1, m + 1): 
        for j in range(1, n + 1): 
            if p[j - 1] == s[i - 1] or p[j - 1] == '.': 
                dp[i][j] = dp[i - 1][j - 1] # Characters match
            elif p[j - 1] == '*': 
                dp[i][j] = dp[i][j - 2] # '*' matches zero occurrences
                if p[j - 2] == s[i - 1] or p[j - 2] == '.': 
                    dp[i][j] = dp[i][j] or dp[i - 1][j] # '*' matches one or more occurrences
    return dp[m][n] # Return whether the entire string matches the entire pattern
# Example usage
s = "aab"
p = "c*a*b"
result = isMatch(s, p)
print(result)  # Output: True
# Example usage
s = "mississippi"
p = "mis*is*p*."
result = isMatch(s, p)
print(result)  # Output: False
```

```python
#11. Container With Most Water
# You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) and (i, height[i]).
# Find two lines that together with the x-axis form a container, such that the container contains the most water.
def maxArea(height): 
    left, right = 0, len(height) - 1 # Initialize two pointers
    max_area = 0 # Variable to store the maximum area found
    while left < right:
        width = right - left # Calculate the width between the two pointers
        current_height = min(height[left], height[right]) # Height is determined by the shorter line
        current_area = width * current_height # Calculate the area
        max_area = max(max_area, current_area) # Update the maximum area found
        if height[left] < height[right]: 
            left += 1 # Move the left pointer to the right
        else:
            right -= 1 # Move the right pointer to the left
    return max_area # Return the maximum area found
# Example usage
height = [1,8,6,2,5,4,8,3,7]
result = maxArea(height)
print(result)  # Output: 49
# Example usage
height = [1,1]
result = maxArea(height)
print(result)  # Output: 1
```

```python
#12. Integer to Roman
# Roman numerals are represented by seven different symbols: I, V, X, L, C, D and M.
# Symbol       Value
# I             1
# V             5
# X             10
# L             50
# C             100
# D             500
# M             1000
def intToRoman(num): 
    val = [
        1000, 900, 500, 400,
        100, 90, 50, 40,
        10, 9, 5, 4,
        1 # Values of Roman numeral symbols, including subtractive combinations
    ]
    syms = [
        "M", "CM", "D", "CD",
        "C", "XC", "L", "XL",
        "X", "IX", "V", "IV",
        "I" # Corresponding Roman numeral symbols
    ]
    roman_numeral = ""
    for i in range(len(val)): 
        while num >= val[i]: 
            num -= val[i] # Subtract the value from num
            roman_numeral += syms[i] # Append the corresponding symbol to the result
    return roman_numeral # Return the final Roman numeral string
# Example usage
num = 1994
result = intToRoman(num)
print(result)  # Output: "MCMXCIV"
# Example usage
num = 58
result = intToRoman(num)
print(result)  # Output: "LVIII"
```

```python
#14. Longest Common Prefix
# Write a function to find the longest common prefix string amongst an array of strings.
def longestCommonPrefix(strs): 
    if not strs: 
        return "" # Return empty string if the input list is empty
    prefix = strs[0] # Start with the first string as the initial prefix
    for s in strs[1:]: 
        while not s.startswith(prefix): 
            prefix = prefix[:-1] # Shorten the prefix until it matches the start of the current string
            if not prefix: 
                return "" # Return empty string if no common prefix is found
    return prefix # Return the longest common prefix found

# Example usage
strs = ["flower", "flow", "flight"]
result = longestCommonPrefix(strs)
print(result)  # Output: "fl"
# Example usage
strs = ["dog", "racecar", "car"]
result = longestCommonPrefix(strs)
print(result)  # Output: ""
```
