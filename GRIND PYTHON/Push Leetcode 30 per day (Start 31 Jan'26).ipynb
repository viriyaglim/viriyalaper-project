```python
#Leetcode 16
#16. 3Sum Closest
#Given an integer array nums of length n and an integer target, find three integers in nums such that the sum is closest to target.
#Return the sum of the three integers.
def threeSumClosest(nums, target):
    nums.sort()
    closest_sum = float('inf')
    n = len(nums)

    for i in range(n):
        left, right = i + 1, n - 1
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum
            if current_sum < target:
                left += 1
            elif current_sum > target:
                right -= 1
            else:
                return current_sum

    return closest_sum
# Example usage:
nums = [-1, 2, 1, -4]
target = 1
result = threeSumClosest(nums, target)
print("Sum of three integers closest to target:", result)  # Output: 2

```

```python
#Leetcode 17 
#Letter Combinations of a Phone Number
#Given a string containing digits from 2-9 inclusive, return all possible letter combinations that the number could represent. Return the answer in any order.
def letterCombinations(digits):
    if not digits:
        return []

    phone_map = {
        '2': 'abc',
        '3': 'def',
        '4': 'ghi',
        '5': 'jkl',
        '6': 'mno',
        '7': 'pqrs',
        '8': 'tuv',
        '9': 'wxyz'
    }

    result = []

    def backtrack(index, path):
        if index == len(digits):
            result.append("".join(path))
            return
        possible_letters = phone_map[digits[index]]
        for letter in possible_letters:
            path.append(letter)
            backtrack(index + 1, path)
            path.pop()

    backtrack(0, [])
    return result
# Example usage:
digits = "23"
result = letterCombinations(digits)
print("Letter combinations of the phone number:", result)  # Output: ["ad","ae","af","bd","be","bf","cd","ce","cf"]
#backtrack function is used for generating all possible combinations by exploring each letter for the current digit and recursively building the combination until the length matches the input digits.

```

```python
#Leetcode 20 
#Valid Parentheses
#Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.
def isValid(s):
    stack = []
    bracket_map = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in bracket_map.values():
            stack.append(char)
        elif char in bracket_map.keys():
            if not stack or stack[-1] != bracket_map[char]:
                return False
            stack.pop()
    return not stack
# Example usage:
s = "()[]{}"
result = isValid(s)
print("Is the parentheses string valid?", result)  # Output: True
s = "(]"
result = isValid(s)
print("Is the parentheses string valid?", result)  # Output: False
```

```python
#Leetcode 36
#Valid Sudoku
#Determine if a 9 x 9 Sudoku board is valid. Only the filled cells need to be validated according to the following rules:
#Each row must contain the digits 1-9 without repetition.
#Each column must contain the digits 1-9 without repetition.
#Each of the nine 3 x 3 sub-boxes of the grid must contain the digits 1-9 without repetition.
def isValidSudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]

    for i in range(9):
        for j in range(9):
            num = board[i][j]
            if num != '.':
                box_index = (i // 3) * 3 + (j // 3)
                if (num in rows[i]) or (num in cols[j]) or (num in boxes[box_index]):
                    return False
                rows[i].add(num)
                cols[j].add(num)
                boxes[box_index].add(num)

    return True
# Example usage:
board = [
  ["5","3",".",".","7",".",".",".","."],
  ["6",".",".","1","9","5",".",".","."],
  [".","9","8",".",".",".",".","6","."],
  ["8",".",".",".","6",".",".",".","3"],
  ["4",".",".","8",".","3",".",".","1"],
  ["7",".",".",".","2",".",".",".","6"],
  [".","6",".",".",".",".","2","8","."],
  [".",".",".","4","1","9",".",".","5"],
  [".",".",".",".","8",".",".","7","9"]
]
result = isValidSudoku(board)
print("Is the Sudoku board valid?", result)  # Output: True
#The isValidSudoku function checks each row, column, and 3x3 sub-box for duplicates using sets, returning False if any duplicates are found and True otherwise.

#Leetcode 37
#Sudoku Solver
#Write a program to solve a Sudoku puzzle by filling the empty cells.
from typing import List

class Solution:
    def solveSudoku(self, board: List[List[str]]) -> None:
        rows = [set() for _ in range(9)]
        cols = [set() for _ in range(9)]
        boxes = [set() for _ in range(9)]
        empties = []

        def box_id(r: int, c: int) -> int:
            return (r // 3) * 3 + (c // 3)

        # initialize sets + list of empty cells
        for r in range(9):
            for c in range(9):
                val = board[r][c]
                if val == '.':
                    empties.append((r, c))
                else:
                    rows[r].add(val)
                    cols[c].add(val)
                    boxes[box_id(r, c)].add(val)

        digits = set("123456789")

        def candidates(r: int, c: int):
            return digits - rows[r] - cols[c] - boxes[box_id(r, c)]

        def backtrack() -> bool:
            if not empties:
                return True

            # MRV: pick empty cell with minimum remaining values
            best_i = -1
            best_cands = None
            min_len = 10

            for i, (r, c) in enumerate(empties):
                cand = candidates(r, c)
                k = len(cand)
                if k == 0:
                    return False
                if k < min_len:
                    min_len = k
                    best_i = i
                    best_cands = cand
                    if k == 1:
                        break

            r, c = empties.pop(best_i)
            b = box_id(r, c)

            for val in best_cands:
                board[r][c] = val
                rows[r].add(val)
                cols[c].add(val)
                boxes[b].add(val)

                if backtrack():
                    return True

                # undo
                board[r][c] = '.'
                rows[r].remove(val)
                cols[c].remove(val)
                boxes[b].remove(val)

            empties.insert(best_i, (r, c))
            return False

        backtrack()
# Example usage:
board = [
  ["5","3",".",".","7",".",".",".","."],
  ["6",".",".","1","9","5",".",".","."],
  [".","9","8",".",".",".",".","6","."],
  ["8",".",".",".","6",".",".",".","3"],
  ["4",".",".","8",".","3",".",".","1"],
  ["7",".",".",".","2",".",".",".","6"],
  [".","6",".",".",".",".","2","8","."],
  [".",".",".","4","1","9",".",".","5"],
  [".",".",".",".","8",".",".","7","9"]
]
solution = Solution()
solution.solveSudoku(board)
print("Solved Sudoku board:")
for row in board:
    print(row)
#The solveSudoku function uses backtracking to fill the Sudoku board by trying each number in empty cells and checking for validity, reverting changes if a dead end is reached.

#Leetcode 38 
#Count and Say
#The count-and-say sequence is a sequence of digit strings defined by the recursive formula:
#countAndSay(1) = "1"
#countAndSay(n) is the way you would "say" the digit string from countAndSay(n-1), which is then converted into a different digit string.
def countAndSay(n):
    if n == 1:
        return "1"

    prev_sequence = countAndSay(n - 1)
    result = []
    count = 1

    for i in range(1, len(prev_sequence)):
        if prev_sequence[i] == prev_sequence[i - 1]:
            count += 1
        else:
            result.append(str(count))
            result.append(prev_sequence[i - 1])
            count = 1

    result.append(str(count))
    result.append(prev_sequence[-1])

    return ''.join(result)
# Example usage:
n = 4
result = countAndSay(n)
print("Count and Say sequence for n =", n, "is:", result)  # Output: "1211"
#The countAndSay function generates the nth term of the count-and-say sequence by recursively building the sequence from the previous term, counting consecutive digits and forming the new term accordingly.

#Leetcode 39
#Combination Sum
#Given an array of distinct integers candidates and a target integer target, return a list of all unique combinations of candidates where the chosen numbers sum to target. You may return the combinations in any order.
def combinationSum(candidates, target):
    result = []

    def backtrack(remaining, combination, start):
        if remaining == 0:
            result.append(list(combination))
            return
        if remaining < 0:
            return

        for i in range(start, len(candidates)):
            combination.append(candidates[i])
            backtrack(remaining - candidates[i], combination, i)
            combination.pop()

    backtrack(target, [], 0)
    return result
# Example usage:
candidates = [2,3,6,7]
target = 7
result = combinationSum(candidates, target)
print("Combinations that sum to target:", result)  # Output: [[7], [2, 2, 3]]
candidates = [2,3,5]
target = 8
result = combinationSum(candidates, target)
print("Combinations that sum to target:", result)  # Output: [[2, 2, 2, 2], [2, 3, 3], [3, 5]]

#The combinationSum function uses backtracking to explore all possible combinations of candidates that sum to the target, allowing for repeated use of the same candidate.

#Leetcode 40 
#Combination Sum II
#Given a collection of candidate numbers (candidates) and a target number (target), find all unique combinations in candidates where the candidate numbers sum to target. Each number in candidates may only be used once in the combination.
def combinationSum2(candidates, target):
    candidates.sort()
    result = []

    def backtrack(remaining, combination, start):
        if remaining == 0:
            result.append(list(combination))
            return
        if remaining < 0:
            return

        for i in range(start, len(candidates)):
            if i > start and candidates[i] == candidates[i - 1]:
                continue
            combination.append(candidates[i])
            backtrack(remaining - candidates[i], combination, i + 1)
            combination.pop()

    backtrack(target, [], 0)
    return result
# Example usage:
candidates = [10,1,2,7,6,1,5]
target = 8
result = combinationSum2(candidates, target)
print("Unique combinations that sum to target:", result)  # Output: [[1,1,6],[1,2,5],[1,7],[2,6]]
candidates = [2,5,2,1,2]
target = 5
result = combinationSum2(candidates, target)

print("Unique combinations that sum to target:", result)  # Output: [[1,2,2],[5]]
#The combinationSum2 function sorts the candidates and uses backtracking to find unique combinations that sum
# to the target, ensuring that each candidate is used only once and skipping duplicates.
#The backtrack function explores combinations by adding candidates to the current combination and recursively searching for valid sums, while avoiding duplicates by checking previous candidates.



```

```python
#Leetcode 126
#Word Ladder II
#Given two words, beginWord and endWord, and a dictionary wordList, return all the shortest transformation sequences from beginWord to endWord, or an empty list if no such sequence exists. Each transformation must change exactly one letter and each transformed word must exist in the word list. Note that beginWord is not a transformed word.

from collections import defaultdict, deque
from typing import List

class Solution:
    def findLadders(self, beginWord: str, endWord: str, wordList: List[str]) -> List[List[str]]:
        wordSet = set(wordList)
        if endWord not in wordSet:
            return []

        L = len(beginWord)

        buckets = defaultdict(list)
        for w in wordSet | {beginWord}:
            for i in range(L):
                buckets[w[:i] + "*" + w[i+1:]].append(w)

        parents = defaultdict(list)
        dist = {beginWord: 0}
        q = deque([beginWord])

        while q and endWord not in dist:
            level_size = len(q)
            patterns_used = set()

            for _ in range(level_size):
                w = q.popleft()
                d = dist[w]

                for i in range(L):
                    pat = w[:i] + "*" + w[i+1:]
                    patterns_used.add(pat)

                    for nei in buckets[pat]:
                        if nei == w:
                            continue

                        if nei not in dist:
                            dist[nei] = d + 1
                            parents[nei].append(w)
                            q.append(nei)
                        elif dist[nei] == d + 1:
                            parents[nei].append(w)

            # clear AFTER the entire level so we don't lose alternative parents
            for pat in patterns_used:
                buckets[pat] = []

        if endWord not in dist:
            return []

        res = []
        path = [endWord]

        def dfs(cur: str):
            if cur == beginWord:
                res.append(path[::-1])
                return
            for p in parents[cur]:
                path.append(p)
                dfs(p)
                path.pop()

        dfs(endWord)
        return res

# Example usage:
solution = Solution()
beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log","cog"]
result = solution.findLadders(beginWord, endWord, wordList)
print("All shortest transformation sequences from beginWord to endWord:", result)  # Output: [["hit","hot","dot","dog","cog"],["hit","hot","lot","log","cog"]]
#The findLadders function finds all the shortest transformation sequences from a beginWord to an endWord using a given word list. It employs a bidirectional breadth-first search (BFS) to explore possible transformations while keeping track of parent-child relationships in a dictionary. Once the shortest paths are identified, it uses depth-first search (DFS) to backtrack from the endWord to the beginWord, constructing all valid transformation sequences. The function returns a list of these sequences.

#Leetcode 127
#Word Ladder
#Given two words, beginWord and endWord, and a dictionary wordList, return the number of words in the shortest transformation sequence from beginWord to endWord, or 0 if no such sequence exists. Each transformation must change exactly one letter and each transformed word must exist in the word list. Note that beginWord is not a transformed word.
from collections import deque
def ladderLength(beginWord, endWord, wordList):
    wordSet = set(wordList)
    if endWord not in wordSet:
        return 0

    L = len(beginWord)
    queue = deque([(beginWord, 1)])
    visited = set([beginWord])

    while queue:
        current_word, level = queue.popleft()

        if current_word == endWord:
            return level

        for i in range(L):
            for c in 'abcdefghijklmnopqrstuvwxyz':
                next_word = current_word[:i] + c + current_word[i+1:]

                if next_word in wordSet and next_word not in visited:
                    visited.add(next_word)
                    queue.append((next_word, level + 1))

    return 0
# Example usage:
beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log","cog"]
result = ladderLength(beginWord, endWord, wordList)
print("Length of the shortest transformation sequence from beginWord to endWord:", result)  # Output: 5
beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log"]
result = ladderLength(beginWord, endWord, wordList)
print("Length of the shortest transformation sequence from beginWord to endWord:", result)  # Output: 0
#The ladderLength function calculates the length of the shortest transformation sequence from a beginWord to an endWord using a given word list. It employs a breadth-first search (BFS) approach, exploring all possible one-letter transformations of the current word. The function maintains a queue to track the current word and its level (number of transformations) and a set to keep track of visited words. If the endWord is reached, it returns the level; otherwise, it returns 0 if no transformation sequence exists.

#Leetcode 128 
#Longest Consecutive Sequence
#Given an unsorted array of integers nums, return the length of the longest consecutive elements sequence. You must write an algorithm that runs in O(n) time.
def longestConsecutive(nums):
    num_set = set(nums)
    max_length = 0

    for num in num_set:
        if num - 1 not in num_set:
            current_num = num
            current_length = 1

            while current_num + 1 in num_set:
                current_num += 1
                current_length += 1

            max_length = max(max_length, current_length)

    return max_length
# Example usage:
nums = [100,4,200,1,3,2]
result = longestConsecutive(nums)
print("Length of the longest consecutive elements sequence:", result)  # Output: 4
nums = [0,3,7,2,5,8,4,6,0,1]
result = longestConsecutive(nums)
print("Length of the longest consecutive elements sequence:", result)  # Output: 9
#The longestConsecutive function calculates the length of the longest consecutive elements sequence in an unsorted array of integers. It uses a set to store the unique numbers and iterates through each number to find the start of a potential sequence (where the previous number is not in the set). It then counts the length of the sequence by checking for consecutive numbers. The function keeps track of the maximum length found and returns it after processing all numbers.

#Leetcode 129
#Sum Root to Leaf Numbers
#You are given the root of a binary tree where each node contains a single digit (0-9). Each root-to-leaf path in the tree represents a number formed by concatenating the digits from the root to the leaf. Return the total sum of all root-to-leaf numbers. A leaf node is a node that does not have any children.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
def sumNumbers(root):
    def dfs(node, current_number):
        if not node:
            return 0

        current_number = current_number * 10 + node.val

        if not node.left and not node.right:
            return current_number

        return (dfs(node.left, current_number) +
                dfs(node.right, current_number))

    return dfs(root, 0)
# Example usage:
root = TreeNode(1, TreeNode(2), TreeNode(3))
result = sumNumbers(root)
print("Total sum of all root-to-leaf numbers:", result)  # Output: 25
root = TreeNode(4, TreeNode(9, TreeNode(5), TreeNode(1)), TreeNode(0))
result = sumNumbers(root)
print("Total sum of all root-to-leaf numbers:", result)  # Output: 1026
#The sumNumbers function calculates the total sum of all root-to-leaf numbers in a binary tree. It uses a depth-first search (DFS) approach to traverse the tree, building the current number by concatenating the digits from the root to the current node. When it reaches a leaf node, it returns the formed number. The function sums up the numbers from all root-to-leaf paths and returns the total sum.

#Leetcode 130
#Surrounded Regions
#Given an m x n matrix board containing 'X' and 'O', capture all regions surrounded by 'X'. A region is captured by flipping all 'O's into 'X's in that surrounded region.
def solve(board):
    if not board or not board[0]:
        return

    rows, cols = len(board), len(board[0])

    def dfs(r, c):
        if r < 0 or r >= rows or c < 0 or c >= cols or board[r][c] != 'O':
            return
        board[r][c] = 'E'
        dfs(r + 1, c)
        dfs(r - 1, c)
        dfs(r, c + 1)
        dfs(r, c - 1)

    for r in range(rows):
        for c in range(cols):
            if (r in [0, rows - 1] or c in [0, cols - 1]) and board[r][c] == 'O':
                dfs(r, c)

    for r in range(rows):
        for c in range(cols):
            if board[r][c] == 'O':
                board[r][c] = 'X'
            elif board[r][c] == 'E':
                board[r][c] = 'O'
# Example usage:
board = [['X','X','X','X'],
         ['X','O','O','X'],
         ['X','X','O','X'],
         ['X','O','X','X']]
solve(board)
print("Board after capturing surrounded regions:")
for row in board:
    print(row)
board = [['X']]
solve(board)
print("Board after capturing surrounded regions:")
for row in board:
    print(row)
#The solve function captures all regions surrounded by 'X' in a given m x n matrix board containing 'X' and 'O'. It uses depth-first search (DFS) to mark 'O's connected to the border as 'E' (escaped). After marking, it iterates through the board to flip all remaining 'O's (which are surrounded) to 'X's and converts 'E's back to 'O's. The function modifies the board in place and does not return anything.
```
